```python
from pyspark.sql import SparkSession
import getpass
username = getpass.getuser()
spark = SparkSession. \
builder. \
config('spark.ui.port', '0'). \
config("spark.sql.warehouse.dir", f"/user/{username}/warehouse"). \
enableHiveSupport(). \
master('yarn'). \
getOrCreate()
```

```python
orders_df = spark.read\
.format("csv")\
.option("header","true")\
.option("inferSchema","true")\
.load("/public/trendytech/orders/*")
```

```python
orders_df.show(truncate=False)
```

```python
orders_df = orders_df.withColumnRenamed("1","order_id")
```

```python
orders_df = orders_df.withColumnRenamed("2013-07-25 00:00:00.0","order_date")
```

```python
orders_df = orders_df.withColumnRenamed("CLOSED","Status")
```

```python
orders_df = orders_df.withColumnRenamed("11599","Customer_id")
```

```python
orders_df.show()
```

#### 1.Top 15 Customer's who placed more number of orders

```python
#orders_df.groupBy("customer_id").count()
result = orders_df.groupBy("customer_id").count().sort("count",ascending=False).limit(15)
```

```python
result.show()
```

```python
reult = spark.sql("select customer_id, count(order_id) as count from orders group by customer_id order by count desc limit 15")
```

```python
result.show()
```
